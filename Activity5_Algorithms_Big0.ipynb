```python
# [CELL 2] Task 7: Algorithm Design (Largest Value)
# Task 6: Problem Decomposition
# Input: Numerical dataset
# Processing: Compare values and retain the largest value
# Output: Largest value

data = [
    5 + NAME_LENGTH,
    20 + SEED_DIGIT,
    8 + NAME_LENGTH,
    15 + ID_SUM
]

largest = data[0]
comparisons = 0

for value in data[1:]:
    comparisons += 1
    if value > largest:
        largest = value

print("Data:", data)
print("Largest:", largest)
print("Comparisons:", comparisons)
```

```python
# [CELL 3] Task 8: Time Complexity (Linear Growth)
def linear_operations(n):
    count = 0
    for _ in range(n):
        count += 1
    return count

for n in [10, 100, 1000, 5000, 10000]:
    print(f"n={n:5d} | operations={linear_operations(n):5d}")
```

```python
# [CELL 4] Task 9: Big O Classification (Logarithmic Growth)
def log_ops(n):
    count = 0
    while n > 1:
        n //= 2
        count += 1
    return count

for n in [10, 100, 1000, 10000]:
    print(f"n={n:4d} | O(1)=1 | O(log n)={log_ops(n):2d} | O(n)={n:5d} | O(n^2)={n*n:7d}")
```

```python
# [CELL 5] Task 10: Space Complexity
def constant_space(n):
    total = 0
    for value in range(n):
        total += value
    return total

def linear_space(n):
    return list(range(n))

for n in [10, 100, 1000, 5000]:
    data = linear_space(n)
    print(f"n={n:5d} | O(1): constant | O(n): {len(data)}")
```

```python
# [CELL 6] Task 11: Best-, Average-, and Worst-Case Analysis
def search_case(data, target):
    for i, value in enumerate(data, 1):
        if value == target:
            return i
    return len(data)

data = [11, 22, 33, 44, 55]

for case, target in {"Best": 11, "Average": 33, "Worst": 99}.items():
    print(f"{case} Case: {search_case(data, target)} comparisons")
```

```python
# [CELL 7] Task 12: Linear Search Implementation
def linear_search(data, target):
    for i, value in enumerate(data):
        if value == target:
            return i, i + 1
    return -1, len(data)

data = list(range(1, 21))
target = SEED_DIGIT + 1

index, comparisons = linear_search(data, target)

print("Target:", target)
print("Index:", index)
print("Comparisons:", comparisons)
```

```python
# [CELL 8] Task 13: Binary Search Implementation
def binary_search(data, target):
    low, high, comparisons = 0, len(data) - 1, 0
    while low <= high:
        comparisons += 1
        mid = (low + high) // 2
        if data[mid] == target:
            return mid, comparisons
        elif data[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1, comparisons

index, comparisons = binary_search(data, target)

print("Target:", target)
print("Index:", index)
print("Comparisons:", comparisons)
```

```python
# [CELL 9] Task 14: Search Algorithm Comparison
data = list(range(1, 101))
target = ((ID_SUM + SEED_DIGIT) % 100) + 1

li, lc = linear_search(data, target)
bi, bc = binary_search(data, target)

print("Target:", target)
print(f"Linear  -> index={li}, comparisons={lc}")
print(f"Binary  -> index={bi}, comparisons={bc}")
print("Reduction:", lc - bc)
```

```python
# [CELL 10] Task 15: Bubble Sort
def bubble_sort(data):
    a = data.copy()
    comparisons, swaps = 0, 0
    for i in range(len(a) - 1):
        swapped = False
        for j in range(len(a) - 1 - i):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1
                swapped = True
        if not swapped:
            break
    return a, comparisons, swaps

data = [
    (ID_SUM * 3) % 100, (SEED_DIGIT * 11) % 100, 
    (NAME_LENGTH * 7) % 100, 42, 17, 89, 31, 56
]

sorted_data, comparisons, swaps = bubble_sort(data)

print("Sorted:", sorted_data)
print("Comparisons:", comparisons)
print("Swaps:", swaps)
```

```python
# [CELL 11] Task 16: Selection Sort
def selection_sort(data):
    a = data.copy()
    comparisons, swaps = 0, 0
    for i in range(len(a) - 1):
        minimum = i
        for j in range(i + 1, len(a)):
            comparisons += 1
            if a[j] < a[minimum]:
                minimum = j
        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            swaps += 1
    return a, comparisons, swaps

sorted_data, comparisons, swaps = selection_sort(data)

print("Sorted:", sorted_data)
print("Comparisons:", comparisons)
print("Swaps:", swaps)
```

```python
# [CELL 12] Task 17: Insertion Sort
def insertion_sort(data):
    a = data.copy()
    comparisons, shifts = 0, 0
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1
        while j >= 0:
            comparisons += 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            shifts += 1
            j -= 1
        a[j + 1] = key
    return a, comparisons, shifts

sorted_data, comparisons, shifts = insertion_sort(data)

print("Sorted:", sorted_data)
print("Comparisons:", comparisons)
print("Shifts:", shifts)
```

```python
# [CELL 13] Task 18: Execution-Time Measurement
import time

def measure(sort_function, data, trials=5):
    times = []
    for _ in range(trials):
        start = time.perf_counter()
        result = sort_function(data)
        times.append(time.perf_counter() - start)
    return result, sum(times) / trials

data_size = 100 + ID_SUM * 10
performance_data = [
    (ID_SUM * 3 + i * 7) % 1000
    for i in range(data_size)
]

bubble_result, bubble_time = measure(bubble_sort, performance_data)
selection_result, selection_time = measure(selection_sort, performance_data)
insertion_result, insertion_time = measure(insertion_sort, performance_data)

print(f"Dataset Size: {data_size}")
print(f"Bubble Sort:    {bubble_time:.8f} s")
print(f"Selection Sort: {selection_time:.8f} s")
print(f"Insertion Sort: {insertion_time:.8f} s")
```

```python
# [CELL 14] Task 19: Sorting Performance Comparison
results = [
    ("Bubble Sort", bubble_result, bubble_time),
    ("Selection Sort", selection_result, selection_time),
    ("Insertion Sort", insertion_result, insertion_time),
]

for name, result, elapsed in results:
    print(f"Algorithm: {name:<18} Comparisons: {result[1]:<12} Movements: {result[2]:<12} Time (s): {elapsed:.8f}")
```
