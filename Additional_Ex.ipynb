```python
import time
import random

# ==========================================
# STUDENT INPUTS & GENERATED VARIABLES
# ==========================================
LAST_NAME = "Royo"
STUDENT_ID = "TUPM-26-1155"
FAVORITE_ARTIST = "Adie"

# Rules implementation
SEED_DIGIT = int(STUDENT_ID[-1])  # 5
ID_SUM = sum(int(char) for char in STUDENT_ID if char.isdigit())  # 20
NAME_LENGTH = len(LAST_NAME)  # 4
ARTIST_LENGTH = len(FAVORITE_ARTIST)  # 4
ARTIST_SUM = sum(ord(char.upper()) - ord('A') + 1 for char in FAVORITE_ARTIST if char.isalpha())  # 23


# ==========================================
# EXERCISE 1: Emergency Dispatch Record Search
# ==========================================
def exercise_1():
    print("==========================================")
    print("EXERCISE 1: Emergency Dispatch Record Search")
    print("==========================================")
    
    # 1. Generate Dataset
    base_val = SEED_DIGIT + ID_SUM  # 5 + 20 = 25
    original_records = sorted([base_val + i * 3 for i in range(10)])  # Unique sorted values
    
    # 2. Rotate Dataset using SEED_DIGIT
    rotation_point = SEED_DIGIT % len(original_records)  # 5
    rotated_records = original_records[rotation_point:] + original_records[:rotation_point]
    
    # 3. Target Incident Calculation
    target_incident = ID_SUM + ARTIST_SUM  # 20 + 23 = 43
    if target_incident not in rotated_records:
        target_incident = rotated_records[SEED_DIGIT % len(rotated_records)]
    
    # 4. Search in Rotated Array (Modified Binary Search)
    def search_rotated(arr, target):
        low, high = 0, len(arr) - 1
        comparisons = 0
        while low <= high:
            mid = (low + high) // 2
            comparisons += 1
            if arr[mid] == target:
                return mid, comparisons
            
            # Left half is sorted
            if arr[low] <= arr[mid]:
                if arr[low] <= target < arr[mid]:
                    high = mid - 1
                else:
                    low = mid + 1
            # Right half is sorted
            else:
                if arr[mid] < target <= arr[high]:
                    low = mid + 1
                else:
                    high = mid - 1
        return -1, comparisons

    # 5. Linear Search Comparison
    def linear_search(arr, target):
        comparisons = 0
        for i, val in enumerate(arr):
            comparisons += 1
            if val == target:
                return i, comparisons
        return -1, comparisons

    pos_rot, comp_rot = search_rotated(rotated_records, target_incident)
    pos_lin, comp_lin = linear_search(rotated_records, target_incident)

    # Output Assessment Data
    print(f"Original Dataset:  {original_records}")
    print(f"Rotated Dataset:   {rotated_records}")
    print(f"Target Incident:   {target_incident}")
    print(f"Rotated Search:    Found at index {pos_rot} with {comp_rot} comparisons")
    print(f"Linear Search:     Found at index {pos_lin} with {comp_lin} comparisons")
    print(f"Time Complexity:   O(log N) for Rotated Binary Search | O(N) for Linear Search")
    print(f"Space Complexity:  O(1)\n")


# ==========================================
# EXERCISE 2: University Enrollment Records
# ==========================================
def exercise_2():
    print("==========================================")
    print("EXERCISE 2: University Enrollment Records")
    print("==========================================")
    
    # 1. Generate sorted datasets
    dept_a = sorted([SEED_DIGIT * 2 + i * ID_SUM for i in range(5)])       # [10, 30, 50, 70, 90]
    dept_b = sorted([ARTIST_LENGTH * 3 + i * NAME_LENGTH for i in range(5)]) # [12, 16, 20, 24, 28]

    # 2. Two-Pointer Consolidation (Merge)
    consolidated = []
    i = j = 0
    comparisons = 0

    while i < len(dept_a) and j < len(dept_b):
        comparisons += 1
        if dept_a[i] <= dept_b[j]:
            consolidated.append(dept_a[i])
            i += 1
        else:
            consolidated.append(dept_b[j])
            j += 1

    while i < len(dept_a):
        consolidated.append(dept_a[i])
        i += 1

    while j < len(dept_b):
        consolidated.append(dept_b[j])
        j += 1

    # Output Assessment Data
    print(f"Department A dataset:  {dept_a}")
    print(f"Department B dataset:  {dept_b}")
    print(f"Consolidated dataset:  {consolidated}")
    print(f"Records before/after:  Before = {len(dept_a) + len(dept_b)} | After = {len(consolidated)}")
    print(f"Comparison count:      {comparisons}")
    print(f"Ordering verification: {'SUCCESS' if consolidated == sorted(consolidated) else 'FAILED'}")
    print(f"Time Complexity:       O(N + M)")
    print(f"Space Complexity:      O(N + M)\n")


# ==========================================
# EXERCISE 3: E-Commerce Order Processing
# ==========================================
def exercise_3():
    print("==========================================")
    print("EXERCISE 3: E-Commerce Order Processing")
    print("==========================================")

    # 1. Sorting Algorithms Implementations
    def bubble_sort(arr):
        data = arr.copy()
        comparisons = movements = 0
        n = len(data)
        start_time = time.perf_counter()
        for i in range(n):
            for j in range(0, n - i - 1):
                comparisons += 1
                if data[j] > data[j + 1]:
                    data[j], data[j + 1] = data[j + 1], data[j]
                    movements += 3  # 3 operations per swap
        exec_time = time.perf_counter() - start_time
        return data, comparisons, movements, exec_time

    def selection_sort(arr):
        data = arr.copy()
        comparisons = movements = 0
        n = len(data)
        start_time = time.perf_counter()
        for i in range(n):
            min_idx = i
            for j in range(i + 1, n):
                comparisons += 1
                if data[j] < data[min_idx]:
                    min_idx = j
            if min_idx != i:
                data[i], data[min_idx] = data[min_idx], data[i]
                movements += 3
        exec_time = time.perf_counter() - start_time
        return data, comparisons, movements, exec_time

    def insertion_sort(arr):
        data = arr.copy()
        comparisons = movements = 0
        start_time = time.perf_counter()
        for i in range(1, len(data)):
            key = data[i]
            j = i - 1
            movements += 1
            while j >= 0:
                comparisons += 1
                if data[j] > key:
                    data[j + 1] = data[j]
                    movements += 1
                    j -= 1
                else:
                    break
            data[j + 1] = key
            movements += 1
        exec_time = time.perf_counter() - start_time
        return data, comparisons, movements, exec_time

    # 2. Generate Initial Input Datasets
    random.seed(SEED_DIGIT)
    base_data = [random.randint(10, 99) for _ in range(15)]
    
    datasets = {
        "Sorted": sorted(base_data),
        "Reverse-ordered": sorted(base_data, reverse=True),
        "Random": base_data.copy()
    }

    algorithms = {
        "Bubble Sort": bubble_sort,
        "Selection Sort": selection_sort,
        "Insertion Sort": insertion_sort
    }

    # 3. Execution & Metrics Collection
    for condition, data in datasets.items():
        print(f"--- Condition: {condition} ---")
        print(f"Input: {data}")
        for alg_name, alg_func in algorithms.items():
            sorted_data, comps, moves, duration = alg_func(data)
            is_valid = sorted_data == sorted(data)
            print(f"  [{alg_name}]")
            print(f"    Comparisons: {comps} | Movements: {moves} | Time: {duration:.6f}s | Verified: {is_valid}")
        print()


# Run all exercises
if __name__ == "__main__":
    exercise_1()
    exercise_2()
    exercise_3()
```
